```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("check-schema")
    .master("local[*]")
    .config("spark.driver.memory", "8g")
    .config("spark.sql.shuffle.partitions", "200")
    .getOrCreate()
)

spark.conf.set("spark.sql.session.timeZone", "UTC")
spark.sparkContext.setLogLevel("WARN")
```

```python
import os

os.chdir("/Users/olaf/bin/ecommerce_project")
print(os.getcwd())


user_daily = spark.read.parquet("/Users/olaf/bin/ecommerce_project/data/processed/user_daily")
session_funnel = spark.read.parquet("/Users/olaf/bin/ecommerce_project/data/processed/session_funnel")

print("===== user_daily schema =====")
user_daily.printSchema()

print("===== session_funnel schema =====")
session_funnel.printSchema()
```
